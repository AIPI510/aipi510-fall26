## Introduction

This notebook designs a hypothesis-driven study to test a client claim and works out how many participants it would need.

**Client:** OfficeOasis Plant Co.

**Claim:** Placing a small desk plant improves employee productivity scores compared to an empty desk.

**Measure:** productivity rating (continuous)

## 1. Study Design

| Design choice | Decision |
| :--- | :--- |
| Participants | Office employees at a standard company. |
| Groups | Treatment = a plant on the desk; control = no plant. |
| Assignment | Random assignment of employees to the two groups. |
| What is measured, and when | Employee productivity rating, measured once at the end of the study period and compared between the two groups. |
| Duration | One fiscal quarter. |

## 2. Hypotheses

- **Null hypothesis (H0):** The plant does not improve average employee productivity compared to an empty desk (productivity is the same or worse with the plant).
- **Alternative hypothesis (H1):** The plant improves average employee productivity compared to an empty desk.
- **One-sided or two-sided:** One-sided — testing for improvement only.

## 3. Statistical Test

**Test:** One-sided independent two-sample t-test.

**Why this measure:** Productivity is a continuous score, and the comparison is the average between two separate groups, plant vs. no plant. It is one-sided because improvement is the only direction being tested.

**Assumptions:**
- **Independence:** each person's score is separate from the others (covered by random assignment of different employees to each group).
- **Roughly normal, similar spread:** scores in each group are roughly normal with similar spread.

## 4. Effect Size From the Literature

**Effect size type:** Standardized mean difference (SMD) — the gap between the two group averages in standard-deviation units.

**Value:** SMD = 0.534 (medium effect). 95% confidence interval 0.167 to 0.901.

**How the value was derived:** The meta-analysis reports academic achievement (grades and exam scores) as the closest outcome to productivity, and it was the one significant, positive result (SMD = 0.534). It is used here as a substitute for employee performance.

**Source:**

- https://pmc.ncbi.nlm.nih.gov/articles/PMC9224521/
- Han, Ke-Tsung et al. "Effects of Indoor Plants on Human Functions: A Systematic Review with Meta-Analyses." _International Journal of Environmental Research and Public Health_ vol. 19,12 7454. 17 Jun. 2022, doi:10.3390/ijerph19127454

## 5. Power Analysis

The power analysis calculates the smallest sample size that provides a good chance of detecting a real effect:

| Input | Value | Reason |
| :--- | :--- | :--- |
| Effect size (SMD) | 0.534 | Same value as the academic literature. |
| Significance level (alpha) | 0.05 | Standard 5% tolerance for a false positive (one-sided). |
| Power | 0.80 | Standard 80% chance of detecting a true effect. |
| Group size ratio | 1.0 | Equal-sized plant and no-plant groups. |

The cell below solves for the sample size with statsmodels' `TTestIndPower`, using the one-sided independent two-sample t-test chosen in Section 3.

```python
import numpy as np
from statsmodels.stats.power import TTestIndPower

# Inputs: expected effect size from the meta-analysis, plus the standard
# 5% false-positive rate and 80% chance of detecting a real effect.
effect_size = 0.534
alpha = 0.05
power = 0.80

# solve_power finds the missing input; leaving nobs1 unset solves for the
# per-group sample size. 'larger' makes it a one-sided test (improvement only).
analysis = TTestIndPower()
n = analysis.solve_power(effect_size=effect_size, alpha=alpha, power=power, alternative="larger")
n_per_group = int(np.ceil(n))   # round up to a whole number of people

print(f"Effect size (SMD): {effect_size}")
print(f"Required sample size per group: {n_per_group}")
print(f"Total sample size: {2 * n_per_group}")
```

**Result.** At the expected effect size (SMD = 0.534), the study needs **45 employees per group, 90 in total**, to have an 80% chance of detecting the improvement at the 0.05 level.

## 6. Conclusion

The study tests whether a desk plant improves employee productivity compared to an empty desk. Employees are randomly split into a plant group and a no-plant group, and each person's productivity rating is measured once at the end of one quarter. Because productivity is a continuous score compared between two separate groups, and the claim only predicts improvement, the plan uses a one-sided independent two-sample t-test. The expected effect size comes from a meta-analysis of indoor plants (SMD = 0.534, the significant academic-achievement result used as a productivity substitute).

Using that effect size with a 0.05 significance level and 80% power, the study needs **45 employees per group (90 total)**.

**Limitation:** the effect size is borrowed from academic-achievement studies rather than workplace productivity, which are not exact fields.
