### <u>Step 1: State null and alternative hypotheses</u>

- H<sub>0</sub>: $\mu$<sub>H</sub> $\leq$ $\mu$<sub>S</sub> + 500
- H<sub>a</sub>: $\mu$<sub>H</sub> $\gt$ $\mu$<sub>S</sub> + 500

where:
- $\mu$<sub>S</sub> is the average daily water consumption (in mL) of standard water bottle users.
- $\mu$<sub>H</sub> is the average daily water consumption (in mL) of HydroHype water bottle users.

<u>Rationale:</u>

The right experiment to conduct is a right-tailed test because HydroHype claims to *increase* daily water consumption with its special water bottle rather than *change it* generally. The `500` in the above hypothesis represents the 500 mL increase in daily water consumption that HydroHype claims its water bottles induces.

### <u>Step 2: Choose a statistical test to perform</u>

This experiment will involve a two-sample, right-sided *t*-test.

<u>Rationale:</u>

- Two-sample: this needs to be a two-sample test because we are comparing two sample means -- the mean of HydroHype water bottle users' daily water consumption and the mean of standard water bottle users' daily water consumption -- as opposed to comparing one mean against some set benchmark. The experiment could also use a paired t-test instead of a two-sample t-test if you had participants first drink from standard water bottles for a period of time and then later had them drink from HydroHype water bottles during a later period of time, then paired observations of average daily water consumption on participant, but without knowing more about the company's access to data, I will go with the simple two-sample approach.
- Right-sided: As discussed above, we are only interested in detecting a positive effect to evaluate HydroHype's claim
- *t*-test: In general, the *t*-distributino is more reliable than the normal distribution because it accounts for the uncertainty inherent in estimating population standard deviation/variance, which will be necessary to compute a test statistic.

### <u>Step 4: Perform power analysis with the identified effect size</u>

The required sample size for our experiment is *n* = 78 (see code below for computation).

Rationale:

The effect size choice used below is selected from a literature review completed in step 3.

The choice of 0.05 for $\alpha$ is based on the fact that this is a standard choice for hypothesis testing. While other common choices include 0.01 and 0.1, these are typically used in scenarios where a false rejection of the null hypothesis has dire consequences or in exploratory settings, respectively. Neither of these situations applies in this commercial setting, so we will use the standard.

We will also keep the standard value of 0.80 for power. While it might be tempting to increase the probability that we avoid failing to detect an effect when one exists, doing so would occur at the cost of an increased required sample size. If stakeholders respond that the amount of data required would cost too much, this is the main lever we can use to adjust the sample size without exposing ourselves to the risk of producing misleading results.

```python
from statsmodels.stats.power import TTestIndPower

# From literature review:
effect_size = 0.40

analysis = TTestIndPower()
n = analysis.solve_power(
    effect_size=effect_size,
    alpha=0.05,
    power=0.80,
    alternative='larger'
)
print(f"Required sample size per group: {n:.0f}")
```
