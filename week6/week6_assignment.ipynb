## Background/Overview: 
 We have been tasked by our client, BrainBoost Gum Co., to run a hypothesis test and a power analysis to calculate the required sample size for the following claim: 

**Chewing our nootropic gum for 10 minutes before an exam improves memory recall scores more than chewing regular mint gum.**

#### 1. Study Design: 
**Experiment Groups:**
- Control Group: Participants in this group chew regular mint gum 10 minutes before a memory test.
- Treatment Group: Participants in this group chew BrainBoost's nootropic gum 10 minutes before a memory test. 

**Participants:** Healthy adults 

**Assignment:** Random Assignment

**Measurement:** Memory recall score

**Duration:** 3 hours

#### 2. Null and Alternate Hypothesis: 
**Null Hypothesis $H_0$:** Chewing nootropic gum for 10 minutes before an exam improves memory recall scores just as much or less than chewing regular mint gum

**Alternate Hypothesis $H_a$:**  Chewing nootropic gum for 10 minutes before an exam improves memory recall scores more than chewing regular mint gum

#### 3. Statiscal Test to be Performed: 

An **independent 2-sample one-sided t test** is the statistical test to be performed. 

**Why one-sided:** It should be a 1 sided t test because we are checking for whether the nootropic gum *improves* the recall gain an individual would get from chewing gum. 

**Key Assumptions**:

- Independence: Since the assignment of an individual to a group is random, we can assume independence.

- Roughly normal improvements: Improvements follow a relatively normal distribution

**Effect size type:** Cohen's d

There wasn't an effect size reported directly, so it was calculated from the reported means and standard errors of means for Long-Term Word Recall. 

| | Mean score | SEM | n |
|---|---|---|---|
| Nootropic gum | 11.5 | 0.75 | 20 |
| Bubble gum | 9.2 | 0.8 | 20 |

**Nootropic gum SD:** $SEM * \sqrt{n} = 0.75 * \sqrt{20} = 3.354$

**Bubble gum SD:** $SEM * \sqrt{n} = 0.8 * \sqrt{20} = 3.578$

**Pooled SD:** $\sqrt{\frac{(3.354)^2 + (3.578)^2}{2}} = 3.468$

**d:** $\frac{11.5 - 9.2}{3.468} = 0.663$







#### 5. Power Analysis: 
We will now run the power analysis using these parameters: 

| Input | Value | Reason |
|---|---|---|
| Effect size (Cohen's d) | 0.663 | Calculated from the reported means and SEMs for Long-Term Word Recall |
| Significance level ($\alpha$) | 0.05 | $\alpha = 0.05$ is a standard value for this parameter|
| Power | 0.80 | 80% chance of detecting a true effect is a standard value for this parameter |
| Group size ratio | 1.0 | Nootropic group and normal gum group are the same size |

```python
import math
from statsmodels.stats.power import TTestIndPower

d = 0.663  # effect size

n = TTestIndPower().solve_power(effect_size=d, alpha=0.05, power=0.80,
                                ratio=1.0, alternative='larger')
print(f"Participants needed per group: {math.ceil(n)}")
```

#### 6. Results and Conclusions
In conclusion, based on the power analysis, we need **29 participants per group** for there to be an 80% chance of observing a statistically significant improvement. This result was obtained by first determining parameters like the significance level ($\alpha = 0.05$) and the power (0.8). The statistical test to conduct (one-tailed t test) was also determined **before** obtaining the effect size. Then, a literature review was done to find an effect size, and if one wasn't found, the effect size (of type Cohen's d) was calculated using the available parameters. Finally, the power analysis was done as presented previously.
