# PetPulse Analysis

Petting a dog for 5 minutes lowers heart rate more effectively than listening to a guided meditation.

Measure: heart rate (continuous)

## Hypotheses

Null Hypothesis: Petting a dog for 5 minutes is just as effective at lowering heart rate than listening to a guided meditation

Alternative Hypothesis: Petting a dog for 5 minutes lowers heart rate more than listening to a guided meditation

## Statistical Test

We will be comparing two distinct samples, and without previewing the data I am not sure about variances, so I would start with Welch's t-Test.

## Effect Size
To measure the effect, Cohen's d seems to be the appropriate measure. To understand what "effectively lowering heart rate" is, I will calculate a mean value for each group and compare the two to reject the null hypothesis or not. 

Other options to be considered were Odds Ratio or Relative Risk Ratio. Because there isn't a pre-defined outcome whose presence/absence in a group would be under test, neither index is appropriate. No measures of association are appropriate when comparing these two isolated groups.


## Literature Referenced:
Dog Pets: https://pubmed.ncbi.nlm.nih.gov/40072925/

Meditation: https://biomedgrid.com/fulltext/volume12/the-effect-of-meditation-on-regulation-of-heart-rate.001736.php


# Study Design Overview

Hypothesis
One-sided, two-group comparison of means
Chosen significance threshold: 0.05 (accepted as an appropriate threshold given the guidance "α = 0.05 is the default in most fields.")

Statistical test
Welch's T-Test = p-value: 0.000646

Outcome measure
Heart Rate (beats per minute)

Effect size estimate
Cohen’s d = 0.4599 (medium effect)

Power analysis result
Statistical Power chosen: 0.80 (accepted per guidance: 80% is the conventional standard — chosen to balance study cost against the risk of missing real findings.)
76 participants per group (total N = 152)

Result: Reject the null hypothesis

```python
# Calculating p-value: Welch's t-Test

# Dog Pets Heart Rate 
dp_mean = 78.6
dp_sd = 11.919
dp_size = 122

# Guided Meditation Heart Rate
gm_mean = 73.75
gm_sd = 8.07
gm_size = 82

# From Workbook: scipy.stats.ttest_ind(group1, group2, equal_var=False), but I am using the studies' provided stats instead of individual observations
from scipy import stats

# Find p-value
t_stat, p_value = stats.ttest_ind_from_stats(
    mean1=dp_mean,
    std1=dp_sd,
    nobs1=dp_size,
    mean2=gm_mean,
    std2=gm_sd,
    nobs2=gm_size,
    equal_var=False
)

# Effect Size
import numpy as np

pooled_sd = np.sqrt(
    ((dp_size - 1) * dp_sd**2 + (gm_size - 1) * gm_sd**2)
    / (dp_size + gm_size - 2)
)

cohens_d = (dp_mean - gm_mean) / pooled_sd

print(f"t-statistic: {t_stat:.4f}")
print(f"p-value: {p_value:.6f}")
print(f"Cohen's d: {cohens_d:.4f}")

from statsmodels.stats.power import TTestIndPower
import math

analysis = TTestIndPower()

n_per_group = analysis.solve_power(
    effect_size=0.46,
    alpha=0.05,
    power=.80,
    ratio=1
)

print(f"{math.ceil(n_per_group)} per group")
```
