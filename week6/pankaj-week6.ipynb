# Week 6 Sync Activity: PetPulse

## Client Claim

Petting a dog for 5 minutes lowers heart rate more effectively than listening to a guided meditation.

**Measure:** Heart rate (continuous)

## 1. Null and Alternative Hypotheses

**Null hypothesis (H₀):** Petting a dog for 5 minutes does not lower heart rate more than listening to a guided meditation.

**Alternative hypothesis (H₁):** Petting a dog for 5 minutes lowers heart rate more than listening to a guided meditation.

**Rationale:** These hypotheses directly test PetPulse's claim by comparing the heart rate response between two groups: participants who pet a dog for 5 minutes and participants who listen to a guided meditation. 

The null hypothesis assumes that petting a dog does not lower heart rate more than guided meditation, while the alternative hypothesis represents PetPulse's claim that it does. The alternative hypothesis is directional because PetPulse specifically claims that dog petting is more effective than guided meditation.

## 2. Statistical Test

I would use an **independent-samples t-test** to compare the mean change in heart rate between the dog-petting group and the guided-meditation group.

**Rationale:** A t-test is appropriate because heart rate is a continuous variable and the study compares the mean change in heart rate between two independent groups. The groups are independent since each participant would be assigned to either the dog-petting group or the guided-meditation group, not both.

Because PetPulse specifically claims that petting a dog lowers heart rate more than guided meditation, I would use a one-sided test.


## 3. Literature and Effect Size



To estimate an effect size, I reviewed a study by von Rotenhan et al. (2026) that examined whether dog-assisted therapy affected heart rate in children undergoing dental treatment. The study included 60 children: 30 received dog-assisted therapy during their dental visit, while the other 30 received regular dental care without a therapy dog.

The researchers measured heart rate before, during, and after treatment.  During treatment, the difference between the dog-assisted therapy group and control group had a reported effect size of approximately **Cohen's d = 0.65**.

I will use **Cohen's d = 0.65** as the estimated effect size for my power analysis.

**Rationale:** This study is relevant because it compares a dog-intervention group with a separate control group and uses heart rate as the physiological outcome. Cohen's d is also the effect size needed for the independent-samples t-test power analysis. I selected the effect size during treatment because it most closely represents the effect of dog interaction while the intervention is occurring.

**Limitation:** The published study involved children undergoing dental treatment and used dog-assisted therapy rather than exactly five minutes of dog petting. According to the study, the children were allowed to interact with the dog if they chose to, so their interaction was not limited to just petting. PetPulse also compares dog petting with guided meditation rather than standard care. Therefore, Cohen's d = 0.65 is being used as an estimate for planning the proposed study rather than an exact expected effect. 

**Source:**
von Rotenhan, A. F., Weiland, S., Kochanowicz, A. M., et al. (2026). Dog-assisted therapy in paediatric dentistry: physiological, psychological, and family-centred benefits in managing child dental anxiety. European Archives of Paediatric Dentistry, 27, 1529–1540.

## 4. Power Analysis

THe power analysis will be used to determine how many participants are needed to test the PetPulse claim.

The inputs for the power analysis are:

- **Effect size (Cohen's d) = 0.65:** Based on the effect size found in the literature.
- **Significance level (α) = 0.05:** This means accepting a 5% chance of rejecting the null hypothesis when it is actually true. I used 0.05 because it is a commonly used significance level
- **Power = 0.80:** This gives the study an 80% chance of detecting an effect of this size if the effect actually exists. I used 0.80 because it is a commonly used power level.
- **Test = One-sided independent-samples t-test:** This matches the statistical test selected for the study and the directional PetPulse claim.
- **Equal group sizes:** The same number of participants will be assigned to the dog-petting and guided-meditation groups.

For this study, heart-rate reduction will be calculated as:
**Heart-rate reduction = heart rate before the activity − heart rate after the activity**

A larger positive value therefore represents a greater reduction in heart rate. Since the PetPulse claim predicts that the dog-petting group will have a greater reduction than the meditation group, the power analysis uses a one-sided test.

```python
from statsmodels.stats.power import TTestIndPower

# Effect size from the published study
effect_size = 0.65

analysis = TTestIndPower()

n = analysis.solve_power(
    effect_size=effect_size,
    alpha=0.05,
    power=0.80,
    alternative="larger"
)

import math

n_per_group = math.ceil(n)

print(f"Required sample size per group: {n_per_group}")
print(f"Total sample size: {n_per_group * 2}")
```

### Result and Interpretation

The power analysis determined that the PetPulse study would need a **minimum of 30 participants per group, or 60 participants total**. Therefore, PetPulse should recruit at least 60 participants for the study, with 30 participants assigned to the dog-petting group and 30 participants assigned to the guided-meditation group.

This sample size would give the study 80% power to detect an effect similar to the Cohen's d of 0.65 found in the literature, using a significance level of 0.05. In other words, 60 participants is the minimum sample size needed for the study based on the assumptions used in this power analysis.
