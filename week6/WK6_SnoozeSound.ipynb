# Week 6: SnoozeSound Power Analysis

This notebook calculates the sample size needed to test whether SnoozeSound produces a greater reduction in sleep onset time than a comparison pillow.

## 1. What are your null and alternative hypotheses?

μ₁ = Average reduction in time to fall asleep for the SnoozeSound group.

μ₂ = Average reduction in time to fall asleep for the comparison pillow group.

Null hypothesis (H₀): μ₁ − μ₂ ≤ 0 minutes

Alternative hypothesis (H₁): μ₁ − μ₂ > 0 minutes

Rationale: I will test whether SnoozeSound produces a greater reduction in time to fall asleep than a comparison pillow. This tests for an improvement, but not whether it reaches the company’s specific 15-minute claim.

## 2. What statistical test will you perform?

Statistical test: A one-sided independent-samples t-test.

Participants will be randomly assigned to the SnoozeSound or comparison-pillow group. I will measure their time to fall asleep before and after the study and compare the average reductions.

Rationale: Sleep onset reduction is a continuous variable, and the groups are independent. I will use α = 0.05 to test whether SnoozeSound produces a greater reduction.

## 3. Explore literature to find an effect size

Mousavi et al. (2026) studied white noise and reported the following reductions in time to fall asleep:

| Group | Mean reduction | Standard deviation | Sample size |

| White noise | 10 minutes | 9 minutes | 30 |

| Control | 2 minutes | 6.5 minutes | 30 |

Mean difference:

10 − 2 = 8 minutes

Pooled standard deviation:

SD = √[(9² + 6.5²) / 2] = 7.85 minutes

Cohen's d:

d = 8 / 7.85 = 1.02

Rationale: This study examines how white noise affects time to fall asleep, so I used its results to estimate an effect size.

Limitation:The study used phone delivered white noise and older adults, not SnoozeSound pillow users.

Reference:

Mousavi, S. A. V., Salehian, M., & Mazlom, S. R. (2026). The effect of white noise on sleep quality and fatigue in community-dwelling older adults: A randomized controlled trial. BMC Geriatrics, 26, Article 883. https://doi.org/10.1186/s12877-026-07311-2

## 4. Conduct a power analysis to determine the required sample size

I will use the book's `TTestIndPower` method with these inputs:

- Cohen's d = 1.02
- Significance level (α) = 0.05
- Statistical power = 0.80
- Test = One-sided independent-samples t-test
- Equal group sizes

```python
import math
from statsmodels.stats.power import TTestIndPower

analysis = TTestIndPower()

n = analysis.solve_power(
    effect_size=1.02,
    alpha=0.05,
    power=0.80,
    alternative="larger"
)

print(math.ceil(n))
```

### Result and Interpretation

The power analysis requires a minimum of 13 participants per group, or 26 participants total.

Rationale: I used Cohen's d = 1.02, α = 0.05, and 80% power. I rounded the calculated sample size up to a whole number.

Limitation: This is a preliminary estimate because the published study did not test the SnoozeSound pillow. The calculation does not establish the company's specific 15-minute claim.

## 5. Justify your decisions

I chose a one-sided independent-samples t-test to compare average reductions in sleep onset time between two randomly assigned, independent groups.

Using α = 0.05, 80% power, and an estimated effect size of 1.02, the power analysis requires 13 participants per group, or 26 total.

The sample size is for detecting an improvement greater than zero minutes, not for testing the specific 15-minute threshold.
