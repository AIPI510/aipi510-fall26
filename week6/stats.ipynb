# Study Design: 

## Claim: 
Users who log their emotions daily using our app report a 25% lower chance of depressive symptoms after 30 days.

Measure: symptoms: yes/no  (binary)


### Step 1: Research Question and Hypothesis
**Research Question:** Does using the MoodMate app to log emotions daily reduce the likelihood of experiencing depressive symptoms after 30 days?

**Null Hypothesis (H0)**: The probability of screening positive for depressive symptoms at day 30 is the same in the MoodMate and control groups.

**Alternative Hypothesis (H1):** The probability of screening positive for depressive symptoms at day 30 differs between the MoodMate and control groups.
**Rationale**: The hypotheses match the chi-square test of independence. We expect a lower symptom probability in the MoodMate group and use the claimed 25% relative risk reduction as the target effect for sample-size planning.

### Step 2: Choosing Statistical Test
We will use a chi-square test of independence to examine whether group assignment (MoodMate vs. control) is associated with depressive symptom status (yes/no) after 30 days.

This test is appropriate because both variables are categorical, and participants will be randomly assigned to two independent groups. Each participant will contribute one outcome. We will check that the expected count in each cell is at least 5; otherwise, we will use Fisher’s exact test.

We will use a significance level of α = 0.05. If the result is statistically significant, we will examine the symptom proportions to determine the direction of the difference. A lower proportion in the MoodMate group would support a reduction in risk, but would not by itself establish a 25% relative reduction.


### Step 3: Outcome Measure:

The primary outcome will be depressive symptom screening status at day 30, assessed using the Patient Health Questionnaire-9 (PHQ-9).
Participants with a PHQ-9 score ≥10 will be classified as screening positive (1), and those with a score <10 as screening negative (0). We will compare the proportion of participants screening positive between the MoodMate and control groups.


Rationale: The PHQ-9 is a validated measure of depressive symptoms, and a predefined cutoff provides a consistent binary outcome that matches the claim’s yes/no measure. Both groups will complete the same assessment at day 30. Screening positive does not constitute a clinical diagnosis.
 

### Step 4: Literature Review for Effect Size:
Everitt et al. (2021) compared a mood-monitoring-only app with a waitlist control over three weeks. They reported no significant between-group differences. This study does not establish that emotion logging alone produces the reduction claimed by MoodMate.

Deady et al. (2022) compared HeadGear, a 30-day behavioral activation and mindfulness app, with a mood-monitoring control app. Depression prevalence over 12 months was 8.0% in the control group and 3.5% in the intervention group. The reported odds ratio was 0.43, with a 95% confidence interval of 0.26–0.70.

For planning, we will assume a control-group symptom probability of 8%, using this study as an indirect reference. MoodMate’s claimed 25% relative risk reduction would correspond to a symptom probability of 6% in the app group. With equal group sizes, these probabilities produce a planning effect size of Cohen’s w = 0.0392, which equals Cramér’s V for our 2×2 table.

Rationale and limitations: This effect size represents the difference we aim to detect, rather than an effect established for MoodMate. The reference study used a different intervention, control condition, follow-up period, and potentially different symptom definition. Therefore, the 8% control probability is uncertain, and we will examine alternative probabilities in the power analysis.

References:
Everitt et al. (2021). https://doi.org/10.1016/j.jad.2021.05.021
Deady et al. (2022). https://doi.org/10.1017/S0033291720002081

### Step 5: Power Analysis:
We will conduct an approximate power analysis for a Pearson chi-square test of independence without continuity correction, using Cohen’s w = 0.0392, a significance level of 0.05, and 80% power. Participants will be allocated equally between the MoodMate and control groups.
A significance level of 0.05 controls the probability of incorrectly rejecting a true null hypothesis. Power of 80% means that the study has approximately an 80% probability of detecting a group difference if the true symptom probabilities are 8% and 6%.

The calculation requires 2,555 participants per group, or 5,110 participants in total, with valid day-30 outcomes.
Sensitivity analysis shows that the required total sample size increases to 8,408 when the assumed control probability is 5%, and decreases to 1,814 when it is 20%. These scenarios retain the same target of a 25% relative risk reduction.

Interpretation: The sample-size estimate depends on the assumed control probability. It provides power to detect a difference when the true relative reduction is 25%; it does not establish that the reduction is at least 25%. Better evidence for the target population and day-30 outcome would strengthen the planning assumptions.

```python
import math
from statsmodels.stats.power import GofChisquarePower

# Main planning assumptions
p_control = 0.08
relative_reduction = 0.25
alpha = 0.05
target_power = 0.80

# Translate the claim into an intervention probability
p_app = p_control * (1 - relative_reduction)

# Cohen's w for a 2x2 table with equal group allocation
p_average = (p_control + p_app) / 2
effect_size = abs(p_control - p_app) / (
    2 * math.sqrt(p_average * (1 - p_average))
)

analysis = GofChisquarePower()

# Use the chi-square power engine with df = 1
required_total = analysis.solve_power(
    effect_size=effect_size,
    nobs=None,
    alpha=alpha,
    power=target_power,
    n_bins=2
)

# Round up and keep the groups equal
n_per_group = math.ceil(required_total / 2)
n_total = n_per_group * 2

print("Cohen's w:", round(effect_size, 4))
print("Participants per group:", n_per_group)
print("Total participants:", n_total)
```

```python
# Repeat the calculation under alternative control probabilities
for p_control in [0.05, 0.08, 0.20, 0.40]:

    p_app = p_control * (1 - relative_reduction)
    p_average = (p_control + p_app) / 2

    effect_size = abs(p_control - p_app) / (
        2 * math.sqrt(p_average * (1 - p_average))
    )

    required_total = analysis.solve_power(
        effect_size=effect_size,
        nobs=None,
        alpha=alpha,
        power=target_power,
        n_bins=2
    )

    n_per_group = math.ceil(required_total / 2)

    print(
        "Control probability:", p_control,
        "App probability:", round(p_app, 4),
        "Total sample size:", n_per_group * 2
    )
```

The main planning scenario requires 5,110 participants in total, with equal allocation between groups. Sensitivity analysis shows that the required sample size varies substantially with the assumed control-group symptom probability. Therefore, the 8% assumption should be treated as an uncertain planning input rather than an established day-30 probability.
