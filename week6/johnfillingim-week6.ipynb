# The Effect Size Files: BrainBoost Gum Co.

**Client claim:** *"Chewing our nootropic gum for 10 minutes before an exam improves memory recall scores more than chewing regular mint gum."*

**Measure:** memory test score (continuous)

This notebook designs a hypothesis-driven study to test that claim. It covers:

1. The null and alternative hypotheses
2. The study design and the statistical test, with the reasons for each
3. A literature review used to choose a planning effect size
4. A power analysis giving the required sample size, with sensitivity checks and a simulation check
5. A summary of every decision and why it was made

## 1. Pinning down the claim

The wording of the claim decides the design, so it's worth reading closely:

| Phrase in the claim | What it means for the study |
|---|---|
| "our nootropic gum ... **more than** regular mint gum" | The comparison is **gum vs. gum**, not gum vs. nothing. The mint-gum group is an *active control*. Any effect of chewing itself (jaw movement, arousal, mint flavour) happens in **both** arms and cancels out. What's left is only the effect of the nootropic ingredients. |
| "for **10 minutes before** an exam" | Gum is chewed for 10 min and **removed**, then the test starts right away. Participants don't chew during the test. |
| "**improves** memory recall scores" | The claim has a direction: the nootropic group scores *higher*. |
| "memory recall scores" | Outcome: number of words correctly recalled on a standard word-list test (continuous, or close to it). |

The first row matters most. Most of the gum-and-memory literature compares gum with **no gum**. That literature measures a different, larger effect than the one BrainBoost is claiming, and section 4 has to account for this.

## 2. Hypotheses

Let $\mu_N$ be the population mean recall score after chewing BrainBoost nootropic gum and $\mu_M$ the population mean after chewing regular mint gum.

- **Null hypothesis** $H_0: \mu_N \le \mu_M$. The nootropic gum does not improve recall compared with mint gum.
- **Alternative hypothesis** $H_1: \mu_N > \mu_M$. The nootropic gum produces higher mean recall than mint gum.

**Why a one-sided test?** The client makes a directional claim, and only a *higher* score supports it. A result showing the nootropic gum *hurts* memory would be worth knowing, but it would not support the claim, so the study doesn't need to spread its α across both tails to test the claim. Section 6 also reports the two-sided sample size, which is the conservative choice if a reviewer or journal asks for two-sided tests. The difference in required *n* is shown, so either choice can be defended.

**Significance level:** α = 0.05. **Target power:** 1 − β = 0.80 (primary), with 0.90 also reported.

## 3. Study design and statistical test

### Design: randomized, double-blind, parallel-group, active-controlled trial

| Element | Choice | Rationale |
|---|---|---|
| Participants | Healthy adults aged 18–30 (e.g., university students), no daily nootropic or high caffeine use | This matches the "before an exam" use case and the student samples in the gum literature. Screening out heavy caffeine users reduces tolerance and withdrawal confounds if the gum contains caffeine. |
| Allocation | 1:1 random allocation to **Nootropic gum** or **Mint gum** | Randomization removes confounding by baseline memory ability. Equal arms give the most power for a fixed total *N*. |
| Blinding | Placebo gum matched for mint flavour, colour, texture, size and packaging; codes held by a third party | The control arm already has mint flavour and chewing, so blinding isolates the ingredient effect. Participants who know their arm could change their effort (expectancy effect). |
| Procedure | Chew assigned gum for 10 min (timed), dispose of it, then start the memory test **immediately** | Onyper et al. (2011) found gum-related benefits only when gum was chewed *before* testing, and they faded within 15–20 min. Following the claim's 10-minutes-before protocol and testing right away gives the claim its fairest test. |
| Outcome | Immediate free recall of a 15-word list (score 0–15), plus delayed recall as a secondary outcome | Word-list recall is the outcome used by Wilkinson et al. (2002), so the results can be compared with that literature. Fixing one **primary** outcome in advance avoids multiple-comparison problems. |
| Standardization | Same time of day, same room, abstain from caffeine for 12 h beforehand, same word list for everyone | Reduces noise (smaller SD means larger standardized *d* for the same raw difference). |

### Statistical test: one-sided **Welch's independent-samples t-test**

- **Two independent groups and a continuous outcome** point to an independent-samples t-test.
- **Welch** rather than Student: Welch's test doesn't assume equal variances. If an active ingredient helps some people but not others, it can widen the spread in the nootropic arm. With equal *n* per arm, Welch loses almost no power when variances *are* equal, so it's the safer default.
- **Normality:** a 0–15 recall score is a bounded count, not truly normal. With about 100+ per arm, the central limit theorem makes the t-test robust. Section 7 checks this by simulating bounded, integer scores. If the data show strong ceiling effects, a **Mann–Whitney U** test will be the pre-registered sensitivity analysis.
- **Effect size metric:** Cohen's *d* = (μ_N − μ_M) / σ_pooled.

### Alternatives considered (and why they weren't chosen as the primary design)

- **Crossover (within-subject) design:** each person tries both gums on separate days. This needs far fewer people because each person is their own control. It was rejected as the primary design because memory tests have strong **practice effects** (people get better at the task) and would need parallel word lists matched for difficulty. Carryover is also a risk if the nootropic has lasting effects. A crossover with counterbalanced order and parallel lists is a reasonable follow-up study.
- **ANCOVA with a baseline recall score:** measuring recall once *before* chewing and adjusting for it removes between-person variance. If baseline and outcome correlate at *r*, the required *n* shrinks by about a factor of (1 − r²) (Borm, Fransen & Lemmens, 2007). Section 6 quantifies this as a possible upgrade. The primary analysis stays the simpler t-test so it matches the assignment and the claim directly.

## 4. Literature review: what effect size is realistic?

There is no published trial of "BrainBoost" gum, so the effect size has to come from related research. The evidence falls into three groups:

### 4a. Gum vs. no gum: the chewing effect (this effect **cancels out** in our design)

| Study | Design / finding | Relevance |
|---|---|---|
| **Wilkinson, Scholey & Wesnes (2002)**, *Appetite* 38(3), 235–236 | Chewing sugar-free gum, compared with no gum, improved immediate word recall by about **24%** and delayed recall by about **36%**. | This is the classic positive finding, but the comparison is gum vs. **no gum**. |
| **Tucha, Mecklinger, Maier, Hammerl & Lange (2004)**, *Appetite* 42(3), 327–329 | Compared no chewing, mimicked chewing, tasteless gum and spearmint gum. **Gum did not improve memory.** | A failed replication. The chewing effect itself is uncertain. |
| **Onyper, Carr, Farrar & Floyd (2011)**, *Appetite* 57(2), 321–328 | 224 undergraduates. Benefits appeared on several measures only when gum was chewed **5 min before** testing (not during), and lasted only about 15–20 min. The press release described the recall advantage as about 2–3 words. | Supports our "chew before, test immediately" protocol. Again, gum vs. no gum. |
| **Miles & Johnson (2007)**, *Appetite* 48(2), 154–158 | Re-examined context-dependent memory effects of gum and did not reliably reproduce them. | More evidence that gum-and-memory effects are fragile. |

**Takeaway:** even the *chewing* effect is inconsistent, and whatever chewing does is present in **both** of our arms. These studies therefore give an **upper bound**, not an estimate, for our effect.

### 4b. The ingredient effect: what the nootropic adds over mint gum

Nootropic gums are usually built around **caffeine**, often combined with **L-theanine** and/or herbal extracts.

| Study | Finding | Relevance |
|---|---|---|
| **Nehlig (2010)**, "Is caffeine a cognitive enhancer?", *J. Alzheimer's Disease* 20(Suppl 1), S85–S94 | A review concluding that caffeine **does not usually improve learning and memory tasks**, especially intentional learning, and does not seem to affect long-term memory. Its benefits come mostly indirectly, through arousal, mood and attention. | Points to a **small** ingredient effect on *recall*. |
| **Camfield, Stough, Farrimond & Scholey (2014)**, *Nutrition Reviews* 72(8), 507–522 | A meta-analysis of 11 randomized controlled trials. It found **moderate** effects of caffeine + L-theanine on **alertness and attention-switching** in the first 2 hours. **Memory was not among the outcomes it pooled.** | The effects are real, but they show up in *attention*, not recall. A memory benefit would be a secondary, smaller effect. |
| **Davidson (2011)**, *Appetite* 57(1), 303–307 | Caffeinated-herbal gum ("Think Gum") improved aspects of memory compared with **standard gum** and with no gum. | This is the **closest analogue** to our claim: an active gum control. However, it is one study, and the **author is affiliated with the manufacturer** (Think Gum LLC), so the reported effect is probably inflated. We use it as evidence that an effect *might* exist, not as an effect size to plan around. |

### 4c. Publication bias and the "winner's curse"

**Open Science Collaboration (2015)**, *Science* 349, aac4716, replicated 100 psychology studies. On average, the replication effect sizes were about **half** the original ones. Small, single-lab, or industry-sponsored studies like those above are the most likely to overstate effects.

### 4d. Choosing the planning effect size

The two approaches below give similar values:

1. **Synthesis of the literature.** The ingredient-only effect on recall should be *smaller* than the inconsistent gum vs. no-gum effect. Caffeine and L-theanine have moderate effects on attention but weak or absent direct effects on memory, and published effects should be roughly halved. This points to a **small effect, around d ≈ 0.2–0.35**.

2. **Smallest effect size of interest (SESOI).** On a 15-word list, an improvement of **1 extra word** is about the smallest gain a student would notice, and the smallest that would justify a premium price. *Assume* (this was not taken from a specific paper) a between-person SD of about **3 words** for immediate recall of a 15-word list. Then:
$$d_{\text{SESOI}} = \frac{1\ \text{word}}{3\ \text{words}} \approx 0.33$$

**Planning effect size: d = 0.30.** This sits inside the literature-based range and slightly below the SESOI, so the study is powered to detect the smallest meaningful benefit (a little conservatively). Section 6 shows how *n* changes for d = 0.20 (sceptical), 0.30 (planning) and 0.50 (the effect the client would like to be true).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from statsmodels.stats.power import TTestIndPower

pd.set_option("display.width", 120)
rng = np.random.default_rng(510)
analysis = TTestIndPower()
```

## 5. Effect size from the smallest effect of interest

This turns the 1-word SESOI into Cohen's *d* under a few assumed SDs. The SD can't be pinned down exactly, so it's shown as a range.

```python
sesoi_words = 1.0
assumed_sd = [2.5, 3.0, 3.5]
pd.DataFrame({
    "assumed SD (words)": assumed_sd,
    "SESOI (words)": sesoi_words,
    "Cohen's d": [round(sesoi_words / s, 3) for s in assumed_sd],
})
```

Across plausible SDs, a 1-word benefit corresponds to *d* ≈ 0.29–0.40. The planning value **d = 0.30** is at the conservative end of this range.

## 6. Power analysis

`statsmodels`' `TTestIndPower.solve_power` solves for the **per-group** sample size of an independent-samples t-test, given *d*, α, power and the direction of the test. (It uses the pooled-variance t-test. With equal *n*, Welch's test has nearly identical power, and section 7 checks this by simulation.)

```python
D_PLAN, ALPHA, POWER = 0.30, 0.05, 0.80

n_primary = analysis.solve_power(effect_size=D_PLAN, alpha=ALPHA, power=POWER,
                                 ratio=1.0, alternative="larger")
n_per_group = int(np.ceil(n_primary))
print(f"Required n per group (d={D_PLAN}, one-sided alpha={ALPHA}, power={POWER}): "
      f"{n_primary:.1f} -> {n_per_group}")
print(f"Total N before attrition: {2 * n_per_group}")
```

### Adjusting for attrition

Some participants won't show up, won't finish the 10-minute chew, or will break the caffeine-abstention rule. Planning for **10% attrition**, we enrol *n* / (1 − 0.10) per group.

```python
ATTRITION = 0.10
n_enrol = int(np.ceil(n_per_group / (1 - ATTRITION)))
print(f"Enrol {n_enrol} per group -> {2 * n_enrol} participants total")
```

### Sensitivity table

Required *n* per group (before attrition) across effect sizes, power levels and test direction:

```python
rows = []
for d, label in [(0.20, "sceptical"), (0.30, "planning"), (0.33, "SESOI (1 word, SD 3)"), (0.50, "client-optimistic")]:
    for alt, alt_label in [("larger", "one-sided"), ("two-sided", "two-sided")]:
        for pw in (0.80, 0.90):
            n = analysis.solve_power(effect_size=d, alpha=ALPHA, power=pw, alternative=alt)
            rows.append({"d": d, "scenario": label, "test": alt_label, "power": pw,
                         "n per group": int(np.ceil(n)),
                         "total N (+10% attrition)": 2 * int(np.ceil(np.ceil(n) / (1 - ATTRITION)))})
sens = pd.DataFrame(rows)
sens.pivot_table(index=["d", "scenario"], columns=["test", "power"], values="n per group").astype(int)
```

```python
sens.query("power == 0.8")[["d", "scenario", "test", "n per group", "total N (+10% attrition)"]].reset_index(drop=True)
```

**How to read the table:** the required sample size grows roughly with 1/d². Halving the assumed effect from 0.5 to 0.25 means about **four times** as many participants. If we had planned around the client's hoped-for d = 0.5, the study would need only about 50 per group. If the true ingredient effect is closer to 0.3, which the literature suggests, that study (about 50 per group) would have only about 44% power and would probably miss it. This is why the effect size has to come from a sceptical reading of the evidence.

### Power curves

```python
# Fixed categorical order (slot 1-3); one series per assumed effect size.
COLORS = {0.20: "#2a78d6", 0.30: "#eb6834", 0.50: "#1baf7a"}
INK, INK_2, GRID = "#0b0b0b", "#52514e", "#e4e3df"

LABEL_AT = {0.20: 260, 0.30: 90, 0.50: 30}  # n where each curve gets its direct label

ns = np.arange(10, 401)
fig, ax = plt.subplots(figsize=(8, 4.8))
for d, c in COLORS.items():
    pw = analysis.power(effect_size=d, nobs1=ns, alpha=ALPHA, alternative="larger")
    ax.plot(ns, pw, color=c, lw=2, label=f"d = {d:.2f}")
    x = LABEL_AT[d]
    ax.annotate(f"d = {d:.2f}", xy=(x, pw[ns == x][0]), xytext=(8, -12), textcoords="offset points",
                color=INK_2, fontsize=9)

ax.axhline(POWER, color=INK_2, lw=1, ls="--")
ax.text(12, POWER + 0.015, "80% power target", color=INK_2, fontsize=9)
ax.axvline(n_per_group, color=COLORS[D_PLAN], lw=1, ls=":")
ax.plot([n_per_group], [POWER], "o", ms=8, color=COLORS[D_PLAN], mec="white", mew=2, zorder=5)
ax.annotate(f"planning design\nn = {n_per_group} per group", xy=(n_per_group, POWER),
            xytext=(12, -38), textcoords="offset points", color=INK, fontsize=9)

ax.set(xlim=(0, 410), ylim=(0, 1.02), xlabel="Participants per group",
       ylabel="Power (one-sided, alpha = 0.05)")
ax.set_title("Power vs. sample size: nootropic gum vs. mint gum", loc="left", color=INK, fontsize=12)
ax.grid(axis="y", color=GRID, lw=0.8)
for s in ("top", "right"):
    ax.spines[s].set_visible(False)
for s in ("left", "bottom"):
    ax.spines[s].set_color(GRID)
ax.tick_params(colors=INK_2)
ax.legend(frameon=False, loc="lower right", labelcolor=INK_2)
plt.tight_layout()
plt.show()
```

### Possible upgrade: adjusting for a baseline score (ANCOVA)

If each participant also completes a different 15-word list *before* chewing, adjusting for that baseline score removes between-person variance. The required *n* is multiplied by about (1 − r²), where *r* is the correlation between baseline and outcome. Memory-test correlations of about 0.5–0.7 are common, but they are assumptions here.

```python
pd.DataFrame([{"baseline-outcome r": r,
               "n per group (ANCOVA, approx.)": int(np.ceil(n_primary * (1 - r**2))),
               "saving vs t-test": f"{r**2:.0%}"} for r in (0.0, 0.3, 0.5, 0.7)])
```

## 7. Simulation check

The analytic power calculation assumes normally distributed data and a pooled-variance t-test. The real study will use **Welch's** test on **integer, bounded** scores (0–15). This section simulates the planned study directly under those conditions:

- Mint arm: mean 7 words, SD 3 (the assumed values from section 4d)
- Nootropic arm: mean 7 + 0.30 × 3 = 7.9 words, SD 3
- Scores rounded to whole words and clipped to 0–15
- One-sided Welch t-test at α = 0.05, n = planned per-group size, 10,000 simulated studies

```python
def simulate_power(n, d, sd=3.0, base=7.0, n_sims=10_000, bounded=True):
    mint = rng.normal(base, sd, size=(n_sims, n))
    noot = rng.normal(base + d * sd, sd, size=(n_sims, n))
    if bounded:
        mint, noot = np.clip(np.rint(mint), 0, 15), np.clip(np.rint(noot), 0, 15)
    p = stats.ttest_ind(noot, mint, axis=1, equal_var=False, alternative="greater").pvalue
    return (p < ALPHA).mean()

sim = pd.DataFrame([
    {"scenario": "normal, continuous", "simulated power": simulate_power(n_per_group, D_PLAN, bounded=False)},
    {"scenario": "integer, clipped 0-15", "simulated power": simulate_power(n_per_group, D_PLAN, bounded=True)},
    {"scenario": "null true (d = 0) -> type I error", "simulated power": simulate_power(n_per_group, 0.0, bounded=True)},
])
sim["analytic"] = [POWER, POWER, ALPHA]
sim.round(3)
```

The simulated power is close to the analytic 0.80, and the false-positive rate under the null is close to 0.05. Using Welch's test and bounded integer scores therefore doesn't break the sample-size calculation. (Clipping and rounding shrink the observed effect slightly, which is one more reason the 10% attrition buffer is useful.)

## 8. Summary of decisions

| Decision | Choice | Rationale |
|---|---|---|
| Comparison | Nootropic gum vs. flavour-matched mint gum | This is the exact comparison in the claim. It cancels the chewing and arousal effect so only the ingredient effect is tested. |
| Hypotheses | H₀: μ_N ≤ μ_M; H₁: μ_N > μ_M | The claim is directional ("improves"). |
| Design | Randomized, double-blind, parallel groups, chew 10 min then test immediately | Randomization controls confounders, blinding controls expectancy, and the timing follows the claim and Onyper et al. (2011). |
| Test | One-sided Welch t-test (Mann–Whitney as sensitivity analysis) | Two independent groups, continuous outcome, no equal-variance assumption. |
| α / power | 0.05 / 0.80 | Standard conventions. Results for 0.90 power and two-sided tests are also reported. |
| Effect size | **d = 0.30** | The ingredient-only effect is expected to be small (Nehlig 2010; Camfield et al. 2014), published effects are inflated (OSC 2015), and it matches the 1-word SESOI. |
| **Sample size** | **139 per group (278 total); enrol 155 per group (310 total) to allow for 10% attrition** | From `TTestIndPower`, confirmed by simulation. |

**Bottom line for BrainBoost:** a convincing test of the claim needs **about 300 participants**. A small study of, say, 30 per group would have only about 31% power to detect a realistic effect (d = 0.30). A "no significant difference" result from a study that small would tell us very little, and a significant result from it would probably overstate the true effect.
