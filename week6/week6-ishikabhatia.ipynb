# Hypothesis

### Null Hypothesis: Receiving a gratitude text does not increase happiness levels compared to people who don't
### Alternative Hypothesis: Receiving a gratitude text does increase happiness levels compared to people who don't

# Statistical Test

### Test: One sided Independent Sample t-Test

### Explanation: One sided because the hypothesis is testing for increase in happiness levels and independent samples t-test because the means are of two groups (control and tested) with no overlap between the two groups

# Literature: Effect Size

### Justification: This study is relevant because it examines gratitude expressed through text messages and outcomes related to happiness.

# Power Analysis 

1. Effect size: 0.4; as cited in literature 
2. Signifance level: 0.05; default alpha value 
3. Power: 0.8, default power value
4. Sample size: calculated below

```python
!python -m pip install statsmodels
```

```python
from statsmodels.stats.power import TTestIndPower

effect_size = 0.4

analysis = TTestIndPower()
n = analysis.solve_power(
    effect_size=effect_size,
    alpha=0.05,
    power=0.80,
    alternative='larger'
)

print(f"Required sample size per group: {n:.0f}")
```

# Conclusion: 

#### Per power testing, the best sample size is 78. However, since we have two groups (control and tested), the total sample size would be 156 (78*2), so it would be 78 participants per group.
